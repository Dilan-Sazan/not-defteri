```python
study_hours = df['Study_Hours_Per_Day'].values   # veya np.array(df['Study_Hours_Per_Day'])
print(type(study_hours))   # numpy.ndarray
print(study_hours.shape, study_hours.ndim, study_hours.dtype)
```

```python
print(study_hours[0])       # ilk öğrencinin çalışma saati
print(study_hours[:5])      # ilk 5 öğrenci
print(study_hours[-1])      # son öğrenci
```

```python
kosul = study_hours > 5
print(study_hours[kosul])          # günde 5 saatten fazla çalışanlar
print(df[study_hours > 5])         # bu öğrencilerin tüm satırlarını da görebilirsin
```

```python
attendance = df['Attendance_Percentage'].values
skor = study_hours * 2 + attendance * 0.5   # kendi uydurduğun bir "başarı skoru" formülü
```

```python
iki_sutun = df[['Study_Hours_Per_Day','Attendance_Percentage']].values
print(iki_sutun.shape)   # (öğrenci_sayısı, 2)
print(iki_sutun[:, 0])   # sadece çalışma saatleri sütunu
```
